# Imports and Configuration

```python
from pathlib import Path
from PIL import Image
from tqdm import tqdm

# Define paths
RAW_DIR = Path("../data/rawdata/pdxe/images")
CLEAN_DIR = Path("../data/procdata/2_drugSensitivity/images_clean")

# Ensure output directory exists
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

# Define cropping parameters (top trim)
# Based on the PDX plots, we remove the top 100 pixels to clear the titles
TOP_TRIM = 100
```

## Processing Function

```python
def process_pdx_figure(image_path, output_path, top_trim):
    """
    Opens a .webp figure, trims the top metadata, and saves as a clean image.
    """
    try:
        with Image.open(image_path) as img:
            width, height = img.size
            
            # Define the box: (left, top, right, bottom)
            # We only change the 'top' coordinate to remove the title
            crop_box = (0, top_trim, width, height)
            
            trimmed_img = img.crop(crop_box)
            
            # Save as PNG or keep as WebP (PNG is often better for ML training)
            # We change extension to .png for standard ML pipelines
            final_output = output_path.with_suffix('.png')
            trimmed_img.save(final_output, "PNG")
            
    except Exception as e:
        print(f"Error processing {image_path.name}: {e}")
```

## Batch Execution

```python
# List all .webp files
image_files = list(RAW_DIR.glob("*.webp"))
print(f"Found {len(image_files)} images to process.")

# Run the loop with a progress bar
for img_path in tqdm(image_files, desc="Trimming Figures"):
    # Output path with the same filename
    out_path = CLEAN_DIR / img_path.stem
    process_pdx_figure(img_path, out_path, TOP_TRIM)

print(f"Done! Cleaned images are located in: {CLEAN_DIR}")
```
